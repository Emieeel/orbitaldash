# N2 Dissociation in Jupyter

This notebook tests the N2 dissociation trajectory workflow inside Jupyter.
It uses the same trajectory builder as the script example and prepares a Dash app object.

```python
from pathlib import Path
import sys

from orbitaldash import launch_app

# Ensure imports from this examples directory work in notebook contexts.
sys.path.insert(0, str(Path.cwd()))
from n2_dissociation_trajectory import build_trajectory
```

```python
# Keep frame count small for a quick notebook test run.
trajectory, active_idx = build_trajectory(n_frames=3, r_min=1.0, r_max=1.4)
print('Frames:', len(trajectory))
print('Active orbitals:', active_idx)
print('PES labels in frame 0:', list(trajectory[0]['metadata']['pes'].keys()))
```

```python
# Build app without starting server (sanity check for notebook execution).
app = launch_app(
    trajectory=trajectory,
    active=active_idx,
    mode='precomputed',
    spacing_options=[0.20],
    cache_dir='./.orbitaldash_cache',
    auto_run=True,
    jupyter=True,
    jupyter_mode='inline',
)

print('App initialized:', app is not None)
```
