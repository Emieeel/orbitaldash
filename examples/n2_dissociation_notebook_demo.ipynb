# N2 dissociation in Jupyter

Shows the CAS(6,6) natural orbitals of N2 (cc-pVDZ) along the N-N bond stretch, embedded in the notebook.
The trajectory comes from `build_trajectory` in `n2_dissociation_trajectory.py`: RHF, then CASCI on a
symmetry-selected valence space, with the active orbitals aligned across frames.

- Requires PySCF and `orbitaldash` (`pip install -e .` from the repo root).
- The first run writes orbital grids to `./.orbitaldash_cache/` (about 30 MB for 20 frames).
- If the app does not embed, use `jupyter_mode='external'` and open the printed URL. If port 8050 is taken, pass `port=8051`.

```python
from pathlib import Path
import sys

from orbitaldash import launch_app

# Make the example script importable whether the kernel starts in examples/ or in the repo root.
examples_dir = Path.cwd() if (Path.cwd() / 'n2_dissociation_trajectory.py').exists() else Path.cwd() / 'examples'
sys.path.insert(0, str(examples_dir))
from n2_dissociation_trajectory import build_trajectory
```

```python
# Bond lengths in Angstrom (equilibrium is about 1.10). Each frame runs RHF + CASCI(6,6).
# n2_dissociation_trajectory.py scans 0.9-3.0 to follow the bond all the way to dissociation.
trajectory, active_idx = build_trajectory(n_frames=20, r_min=0.8, r_max=1.4)
print('Frames:', len(trajectory))
print('Active orbitals:', active_idx)
print('PES labels in frame 0:', list(trajectory[0]['metadata']['pes'].keys()))
```

```python
# Precomputes the orbital grids on the first run, then embeds the app in the notebook.
app = launch_app(
    trajectory=trajectory,
    active=active_idx,
    mode='precomputed',
    spacing_options=[0.20],
    cache_dir='./.orbitaldash_cache',
    auto_run=True,
    jupyter=True,
    jupyter_mode='inline',
)
```
